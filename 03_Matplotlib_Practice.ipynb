```python
import numpy as np
import matplotlib.pyplot as plt

# Amplitude
A = 5

# Generate x values from 0 to 2π
x = np.linspace(0, 2 * np.pi, 100)

# Generate sine wave
y = A * np.sin(x)

# Plot the sine wave
plt.plot(x, y)

# Labels and title
plt.xlabel("x")
plt.ylabel("Amplitude")
plt.title("Sine Wave")

# Show grid
plt.grid()

# Display graph
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt

# Generate random amplitude between 1 and 10
A = np.random.randint(1, 11)

print("Random Amplitude:", A)

# Generate x values
x = np.linspace(0, 100, 1000)

# Generate sine wave
y = A * np.sin(x)

# Plot the sine wave
plt.plot(x, y)

plt.xlabel("x")
plt.ylabel("Amplitude")
plt.title("Sine Wave with Random Amplitude")
plt.grid()

plt.show()
```

```python
import matplotlib.pyplot as plt

# Height groups
height = ['140-149', '150-159', '160-169', '170-179', '180-189', '190-199']

# Number of people
people = [5, 12, 20, 25, 15, 5]

# Create horizontal bar graph
plt.barh(height, people)

plt.xlabel("Number of People")
plt.ylabel("Height (cm)")
plt.title("Height-wise People Graph")

plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt

# Generate values from 0 to 2π
t = np.linspace(0, 100, 100)

# X-axis = sine wave
x = np.sin(t)

# Y-axis = cosine wave
y = np.cos(t)

# Plot graph
plt.plot(x, y)

plt.xlabel("Sine Wave")
plt.ylabel("Cosine Wave")
plt.title("Sine vs Cosine")

plt.grid()
plt.show()
```
