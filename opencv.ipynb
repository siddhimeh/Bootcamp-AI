```python
import cv2
import numpy as np
import matplotlib as plt
```

```python
from pathlib import Path

image_path = Path("image.jpg")
if not image_path.exists():
    raise FileNotFoundError(f"Image not found: {image_path.resolve()}")

img = cv2.imread(str(image_path))
if img is None:
    raise ValueError(f"OpenCV could not read the image: {image_path.resolve()}")

print(f"Loaded image: {img.shape[1]} x {img.shape[0]} pixels")
```

```python
print(type(img))
```

```python
cv2.imshow("image",img)
cv2.waitKey(0)
cv2.destroyAllWindows()
```

```python
print(img.shape)
```

```python
img_resize = cv2.resize(img,(203,300))
```

```python
cv2.imshow("manya",img_resize)
cv2.waitKey(0)
cv2.destroyAllWindows()
```

```python
print(img_resize.shape)
```

```python
img_flip = cv2.flip(img,1)
```

```python
cv2.imshow("flipped Image",img_flip)
cv2.waitKey(0)
cv2.destroyAllWindows()
```

```python
img_crop = img[100:300,200:500]
```

```python

cv2.imshow("Cropped Image",img_crop)
cv2.waitKey(0)
cv2.destroyAllWindows()
```

```python
cv2.imwrite(r"croppedImg16.jpg",img_crop)
```
